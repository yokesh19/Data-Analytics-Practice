```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
tel=pd.read_csv("C:\\Data Analytics\\archive\\Telecom+Customer+Churn\\telecom_customer_churn_revised.csv")
X=pd.get_dummies(tel[['Gender','Age','City','Phone Service','Internet Service', 'Internet Type','Contract','Tenure in Months','Number of Referrals','Number of Dependents']],drop_first=True)
y=tel['Customer Status']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.25,random_state=42)
model=DecisionTreeClassifier(max_depth=6,min_samples_leaf=6,random_state=42)
model.fit(X_train,y_train)
train_pred=model.predict(X_train)
test_pred=model.predict(X_test)
print("Accuracy_Score_train:",accuracy_score(y_train,train_pred))
print("Accuracy_Score_test:",accuracy_score(y_test,test_pred))
results=X_test.copy()
results['Customer ID'] = tel.loc[X_test.index, 'Customer ID'].values
results['Actual Output']=y_test.values
results['Predicted Output']=test_pred
results.to_csv("C:\\Data Analytics\\archive\\Telecom+Customer+Churn\\telecom_ML_Predictions.csv",index=False)
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
tel=pd.read_csv("C:\\Data Analytics\\archive\\Telecom+Customer+Churn\\telecom_customer_churn_revised.csv")
X=pd.get_dummies(tel[['Gender','Age','City','Phone Service','Internet Service', 'Internet Type','Contract','Tenure in Months','Number of Referrals','Number of Dependents']],drop_first=True)
y=tel['Customer Status']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.25,random_state=42)
model=RandomForestClassifier(n_estimators=200,random_state=42)
model.fit(X_train,y_train)
train_pred=model.predict(X_train)
test_pred=model.predict(X_test)
print("Accuracy_Score_train:",accuracy_score(y_train,train_pred))
print("Accuracy_Score_test:",accuracy_score(y_test,test_pred))
```
